## 0.3 Setting up the folder structure

```python
# Setup the repository structure for this project
folders = [
    'src/bonefracture',   # Contains the different python functions involved in processing the data, evaluating and training the model
    'data',               # Contains the raw dataset of BFD
    'datasets',           # Contains further datasets (augmented, enriched with other datasets) (EMPTY)
    'metrics',            # Contains functions for metrics (EMPTY)
    'notebooks',          # Contains the notebooks that were used in this project
    'models',             # Contains the trained models in the different yolov8 runs
    'test'                # For notebooks and testing ground

]
for f in folders:
    os.makedirs(f, exist_ok=True)

open('src/bonefracture/__init__.py', 'a').close()

for f in ['data_processing.py', 'train.py', 'evaluate.py', 'inference.py']:
  open(f'src/bonefracture/{f}', 'a').close()

print("Structure created: ")

```
